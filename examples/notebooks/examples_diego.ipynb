### Quick simple s.a. optimization example 
System under consideration:
$$\begin{matrix}\dot{x}(t) = A x(t) + A_1 x(t-\tau) + Bu(t)\\ y = Cx(t) \end{matrix}$$
with
$$A_0 = \left[ \begin{matrix}1 &3 \\ 2 &4\end{matrix} \right] \quad A_1 = \left[ \begin{matrix}1 &2 \\ 1 &4\end{matrix} \right] \quad B = \left[ \begin{matrix}1 \\ 1 \end{matrix} \right] \quad u = K x(t) \quad K = \left[ \begin{matrix}k_1 & k2\end{matrix}\right] \quad C = \left[ \begin{matrix}1 & 0\\ 0 & 1 \end{matrix} \right]$$

```python
import numpy as np
import matplotlib.pyplot as plt
import tdspy as tds
```

```python
A0 = np.array([[1, 3], [2, 4]])
A1 = np.array([[1, 2], [1, 4]])
A = [A0, A1]
B = np.array([[1], [1]])
C = np.eye(2)
k = np.array([1, 1]) # Initial guess for the optimal gains
tau = np.array([0, 1])
```

```python
rdde = tds.RDDE(A=A, hA=tau, B1=B, C1 = C)
tds.spectral_abscissa(rdde)
```
